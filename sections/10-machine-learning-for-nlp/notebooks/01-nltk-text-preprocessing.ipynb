## Tokenization

```python
from nltk.tokenize import word_tokenize, sent_tokenize

paragraph = """Natural Language Processing is amazing. It helps machines understand text.
Krish Naik makes it easy to learn!"""

print("sentences:", sent_tokenize(paragraph))
print("words:", word_tokenize(paragraph))
```

## Stemming vs Lemmatization

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

words = ["history", "finally", "studying", "better", "eating"]

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

print("word        stemming   lemmatization")
for w in words:
    print(f"{w:<12}{stemmer.stem(w):<11}{lemmatizer.lemmatize(w, pos='v')}")
```

## Stopwords

```python
from nltk.corpus import stopwords

sw = set(stopwords.words("english"))
tokens = word_tokenize(paragraph.lower())
filtered = [t for t in tokens if t.isalnum() and t not in sw]
print(filtered)
```

## POS Tagging and Named Entity Recognition

```python
from nltk import pos_tag, ne_chunk

tags = pos_tag(word_tokenize("Krish Naik teaches NLP in India."))
print(tags)
print(ne_chunk(tags))
```
