# Stemming

**What?** Chopping word endings with hand-written rules to merge word forms into one *stem*: `playing / played / plays -> play`. The stem does **not** have to be a real dictionary word (`studies -> studi`).

**Why learn it?** In the upcoming BOW/TF-IDF notebooks every word form becomes a separate feature: `connect`, `connected`, `connection` = 3 features for one concept. Stemming collapses them -> smaller vocabulary, fewer sparse features, better generalization. Search engines do the same so a query for `running shoes` also matches *run*.

No NLTK download needed - stemmers are pure rule-based algorithms.

## PorterStemmer - the classic

**What?** The oldest and most widely used stemmer (1980); applies ~60 suffix-stripping rules in stages.

**Why start here?** It is the default reference point every other stemmer is compared against.

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "playing", "played", "plays", "player",        # one concept, four forms
    "studies", "studying", "studied",              # same family, tricky endings
    "easily", "happiness", "happier", "unhappy",   # adverbs / adjectives
    "connected", "connection", "connections",
    "computers", "running", "runs",
    "ran", "dogs", "cats", "mice",                 # irregular forms
    "was", "were", "better", "caring",
]

for w in words:
    print(f"{w:<12} -> {stemmer.stem(w)}")
```

Read the output critically - stemming is crude by design:

- stems are often non-words: `studies -> studi`, `happiness -> happi`, `easily -> easili`
- irregular forms are missed: `ran`, `mice`, `better`, `were` stay unchanged (rules only strip suffixes)
- `player` keeps its suffix - the rules cannot know it derives from *play*

Keep these failures in mind: they are exactly why lemmatization exists (next notebook).

## RegexpStemmer - your own rules

**What?** Strips whatever suffix matches your regex.

**Why?** Full control for domain-specific text (product codes, log messages). It is also the cheapest way to see that stemming is *just rules*: without language knowledge they break - `king` loses a non-suffix `ing` and `caring` loses the `e` it should keep.

```python
from nltk.stem import RegexpStemmer

stemmer = RegexpStemmer(r"ing$|ed$|s$")  # strip a trailing "ing", "ed" or "s"

for word in ["playing", "played", "cats", "dogs", "caring", "king"]:
    print(f"{word:<10} -> {stemmer.stem(word)}")
```

## SnowballStemmer - Porter, improved

**What?** A modern refinement of Porter (a.k.a. Porter2), available for many languages.

**Why?** Slightly more accurate - compare `was -> was` (Snowball) vs `was -> wa` (Porter) below - and multilingual. If you want a stemmer at all, prefer this one.

```python
from nltk.stem import PorterStemmer, SnowballStemmer

porter = PorterStemmer()
snowball = SnowballStemmer("english")

words = [
    "playing", "played", "plays", "player",        # one concept, four forms
    "studies", "studying", "studied",              # same family, tricky endings
    "easily", "happiness", "happier", "unhappy",   # adverbs / adjectives
    "connected", "connection", "connections",
    "computers", "running", "runs",
    "ran", "dogs", "cats", "mice",                 # irregular forms
    "was", "were", "better", "caring",
]

print(f"{'word':<12}{'porter':<10}{'snowball'}")
for w in words:
    print(f"{w:<12}{porter.stem(w):<10}{snowball.stem(w)}")
```

## Takeaways

- Stemming = fast, crude normalization of word forms; output may not be a real word.
- Porter = classic default, Snowball = improvement, Regexp = DIY for special domains.
- Failures to remember: non-word stems, over-stemming (`caring -> car` with naive rules), irregulars missed (`ran`, `mice`, `better`).
