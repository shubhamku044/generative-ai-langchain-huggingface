# Tokenization

Part 1 of 3 - next: `01b-stemming.ipynb`

**What?** Splitting raw text into smaller units called *tokens*. A token is usually a word, but a sentence can also be a token.

**Why learn it?** Every NLP model downstream (BOW, TF-IDF, word2vec, BERT) consumes tokens, not paragraphs. Whatever gets tokenized wrongly here becomes a wrong feature there - this is the foundation of the whole pipeline.

Key terms:
- **Corpus** - the whole collection of text (our paragraph below; could be thousands of documents)
- **Document** - one piece of text (here: one sentence)
- **Vocabulary** - the set of *unique* tokens

## Setup

One-time NLTK data download (cached in `~/nltk_data`).

```python
import nltk

nltk.download("punkt_tab")  # pre-trained tokenizer data
```

## The corpus

One paragraph that we will reuse in every demo.

```python
paragraph = "Yesterday afternoon, a curious traveler discovered an abandoned lighthouse near the rocky coastline, where hundreds of seabirds circled above the waves and salty winds carried the smell of distant rain. Inside the dusty building, she found three wooden boxes, an ancient map, several handwritten letters, and a broken pocket watch that had stopped working sometime in 1947. Although the sky was becoming darker, she decided to investigate the mysterious objects before returning to the village. Suddenly, a small orange cat appeared from behind a staircase and knocked over a glass bottle, revealing a tiny piece of paper hidden beneath it. The note contained a strange message: 'When the moon rises above the northern hills, follow the silver path until you reach the place where time stands still.' Confused but fascinated, the traveler carefully folded the paper, placed it inside her backpack, and walked toward the harbor, wondering whether the discovery was merely an elaborate prank or the beginning of an extraordinary adventure. Meanwhile, several fishermen were repairing their nets, children were collecting colorful shells along the beach, and an elderly woman was selling freshly baked bread from a small wooden cart. Nobody seemed to notice the unusual events unfolding near the lighthouse, even when a distant bell rang thirteen times despite the clock tower showing exactly six o'clock. By midnight, thick clouds had covered the stars, the streets were completely deserted, and the mysterious silver path had appeared across the wet cobblestones, leading directly toward a forgotten garden surrounded by towering trees, rusted iron gates, and strangely glowing flowers."

print("characters:", len(paragraph))
```

## Sentence tokenization

**What?** `sent_tokenize` splits a paragraph into a list of sentences.

**Why?** Many tasks work per sentence (sentiment, summarization, translation). Splitting on `.` naively breaks on abbreviations, decimals and quotes - Punkt is trained to handle these.

```python
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(paragraph)  # each sentence = one document

print("no. of sentences:", len(sentences))
print(sentences)
```

## Word tokenization

**What?** `word_tokenize` splits text into words *and* punctuation (`'` and `.` become their own tokens; `o'clock` and `1947` stay intact).

**Why?** Words are the unit we count and vectorize later. Note the vocabulary is much smaller than the token count - repeated forms collapse.

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(paragraph)

print("total tokens:", len(tokens))
print("vocabulary size (unique tokens):", len(set(tokens)))
tokens[:15]
```

## wordpunct_tokenize - punctuation as its own token

**What?** Splits purely on whitespace and punctuation: `don't` -> `don`, `'`, `t`.

**Why?** Dumber but perfectly predictable. Useful for noisy text (tweets, logs) or when you want punctuation isolated so you can filter it - at the cost of more tokens.

```python
from nltk.tokenize import wordpunct_tokenize

wp_tokens = wordpunct_tokenize(paragraph)

print("word_tokenize:      ", len(tokens), "tokens")
print("wordpunct_tokenize: ", len(wp_tokens), "tokens")
wp_tokens[:15]
```

## TreebankWordTokenizer - the class API

**What?** A class applying the Penn Treebank conventions - the rules `word_tokenize` is based on (it uses `NLTKWordTokenizer`, a slightly improved Treebank variant, after sentence-splitting).

**Why?** Use the class when you want Treebank rules on raw text without the sentence-splitting step, or when you need a reusable/configurable tokenizer object.

```python
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()
treebank_tokens = tokenizer.tokenize(paragraph)

print("treebank tokens:", len(treebank_tokens))
treebank_tokens[:15]
```

## Takeaways

- Pipeline: **corpus -> documents (sentences) -> tokens -> vocabulary**
- `word_tokenize` is the sensible default; `wordpunct_tokenize` when you want simple and predictable; `TreebankWordTokenizer` for the raw class API.
