# Bag of Words, N-grams, TF-IDF

Lectures 11–20: converting text to numbers. Built on scikit-learn's `CountVectorizer` / `TfidfVectorizer`.

```python
corpus = [
    "Machine learning is fun and powerful",
    "Deep learning is a subset of machine learning",
    "NLP deals with text and language processing",
]
```

## Bag of Words (unigrams)

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()  # try lowercase=False, or stop_words='english'
X = cv.fit_transform(corpus)
print("vocabulary:", cv.get_feature_names_out())
print(X.toarray())  # sparse in real life: shape (n_docs, vocab_size)
```

## N-grams (captures word order)

```python
cv_ng = CountVectorizer(ngram_range=(1, 2))  # unigrams + bigrams
X_ng = cv_ng.fit_transform(corpus)
print("vocab with bigrams:", cv_ng.get_feature_names_out())
```

## TF-IDF

$TF(t,d) = \frac{\text{count of } t \text{ in } d}{\text{total terms in } d}$  
$IDF(t) = \log\frac{N}{\text{df}(t)}$  (sklearn uses a smoothed variant)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
X_t = tfidf.fit_transform(corpus)
print("vocabulary:", tfidf.get_feature_names_out())
print(X_t.toarray().round(3))
```
